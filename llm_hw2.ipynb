## Experimenting with Prompts

Large language models (LLMs) are pre-trained on vast amounts of data to generate novel content, including text, images, and video. To generate the desired response, you should feed the input to a model as a prompt. It is important to provide a context-based, clear, and precise prompt to guide the model and get the desired output. The process of designing effective prompts to generate better and desired responses is called prompt engineering.

In this lab, you will learn the basics of prompt engineering and experiment with simple prompts to control how a model interprets and generates desired and coherent responses.

### Learning Objectives:
* Design a prompt structure to write concise prompts based on requirements.
* Write unambiguous and context-based prompts to generate the desired output.
* Write prompts in different formats, such as questions, statements, and instructions.
* Let's get started prompting and experimenting with foundation models.

## Exercise 3: Prompts in different formats

Prompts can be presented in different formats, such as questions, statements, or instructions. Let's experiment with how prompts in different formats can yield different results.

First, let's write a prompt in question format. Enter the prompt

```python
What are the benefits of water reservoirs in a detailed paragraph?
```

### Prompt 1 (Question format):

```python
What are the benefits of water reservoirs in a detailed paragraph?
```

### Response 1:

```python
Water reservoirs provide numerous essential benefits to communities and ecosystems. They serve as crucial water storage facilities, collecting and preserving water during rainy seasons for use during dry periods, thus ensuring a reliable water supply for drinking, agriculture, and industry. Reservoirs play a vital role in flood control by regulating river flow and preventing downstream flooding. They support irrigation systems that enable agricultural production in otherwise arid regions. Additionally, many reservoirs facilitate hydroelectric power generation, providing clean, renewable energy. They also create recreational opportunities such as fishing, boating, and wildlife habitats, while contributing to groundwater recharge and environmental conservation efforts.
```

Next, write the prompt in statement format. 

```python
Discuss the benefits of utilizing water reservoirs.
```

### Prompt 2 (Statement format):

```python
Discuss the benefits of utilizing water reservoirs.
```

### Response 2:

```python
The utilization of water reservoirs offers multiple significant advantages across various sectors. These artificial water bodies provide reliable water storage solutions that ensure consistent supply for municipal, agricultural, and industrial needs throughout the year. They contribute substantially to flood management by controlling river discharge during heavy rainfall events. The agricultural sector benefits immensely from reservoir-supported irrigation systems that enable crop cultivation in water-scarce regions. In the energy sector, reservoirs facilitate hydropower generation, offering a sustainable electricity source. Furthermore, they enhance recreational activities, support aquatic ecosystems, promote tourism, and assist in maintaining environmental flow requirements in river systems.
```

Further, write the prompt in an instruction format.

List the top five benefits of water reservoirs.

### Prompt 3 (Instruction format):

```python
List the top five benefits of water reservoirs.
```

### Response 3:

```python
1. Water Supply: Ensures reliable drinking water and supports agricultural irrigation
2. Flood Control: Regulates river flow and prevents downstream flooding
3. Hydropower Generation: Produces clean, renewable electricity
4. Recreation: Provides opportunities for fishing, boating, and tourism
5. Environmental Support: Maintains ecosystems and supports biodiversity
```

You can review and analyze the response generated by each prompt to identify which prompt format works best to accomplish the desired task. For example, the instruction format would work well for a summarization task. For text completion, the statement format would be suitable. When you want to extract a specific entity from a statement, the question format would be most suitable.

### Сравнение:

Каждый формат промпта дает различную структуру ответа:
- **Вопрос** производит развернутый абзац с подробным объяснением
- **Утверждение** создает более формальный, аналитический текст, подходящий для отчетов
- **Инструкция** генерирует краткий структурированный список, идеальный для быстрого ознакомления

Формат инструкции наиболее эффективен для задач суммирования, тогда как вопрос лучше подходит для получения развернутых объяснений.
